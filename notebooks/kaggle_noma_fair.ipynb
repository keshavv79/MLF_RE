# NOMA v3-FAIR (M=4) pipeline — fig4/5/8 + eval & ablation
Attach the NOMA-fair dataset, then **Save Version → Save & Run All (Commit)**.

```python
import os, shutil, glob
from pathlib import Path
WORK = Path('/kaggle/working'); WORK.mkdir(exist_ok=True); os.chdir(WORK)
inps = sorted(glob.glob('/kaggle/input/*'))
assert inps, 'No dataset attached! Add your M=4 NOMA-fair dataset.'
DATA = inps[0]; print('Input:', DATA)
need_py = ['train_dl_v3_easy.py', 'train_noma_fair_maml.py', 'train_noma_fair_noris.py', 'compare_noma_fair.py', 'aggregate_fig5.py', 'aggregate_fig8.py', 'render_fair_plots.py']
need_mat = ['ISAC_RIS_NOMA_channels_v3_fair.mat', 'ISAC_RIS_NOMA_channels_v3_fair_N16.mat', 'ISAC_RIS_NOMA_channels_v3_fair_N32.mat', 'ISAC_RIS_NOMA_channels_v3_fair_N64.mat', 'ISAC_RIS_NOMA_channels_v3_fair_TEST.mat', 'ISAC_NOMA_channels_fair_noris.mat']
found = {}
for f in glob.glob(os.path.join(DATA,'**','*'), recursive=True):
    b = os.path.basename(f)
    if b in need_py or b in need_mat: shutil.copy(f, WORK/b); found[b]=True
print('Copied :', sorted(found))
print('MISSING:', [f for f in need_py+need_mat if f not in found] or 'none')
for d in ['fig4','fig5','fig8/ris','fig8/no_ris','no_ris']:
    (WORK/d).mkdir(parents=True, exist_ok=True)
for agg,dst in [('aggregate_fig5.py','fig5'),('aggregate_fig8.py','fig8')]:
    if (WORK/agg).exists(): shutil.copy(WORK/agg, WORK/dst/agg)
nor = 'ISAC_NOMA_channels_fair_noris.mat'
if (WORK/nor).exists(): shutil.copy(WORK/nor, WORK/'no_ris'/nor)
import torch
print('CUDA:', torch.cuda.is_available())
EPOCHS=40; BATCH=64; LR='1e-4'
print('EPOCHS',EPOCHS,'BATCH',BATCH,'LR',LR)
```

## fig4 — batch x lr sweep (Anshul-style: batch{32,64,128} x lr{1e-4,1e-5})
The canonical b=64,lr=1e-4 run is saved as policy_noma_fair_best.pt (used by eval/fig5/fig8).

```python
import subprocess, sys
def run(cmd):
    print('>>',' '.join(str(c) for c in cmd)); sys.stdout.flush()
    subprocess.run([str(c) for c in cmd], check=True)

FIG4=[(b,lr) for b in (32,64,128) for lr in ('1e-4','1e-5')]
for (b,lr) in FIG4:
    out = 'policy_noma_fair_best.pt' if (b==64 and lr=='1e-4') else f'fig4/fig4_b{b}_lr{lr}.pt'
    run(['python','-u','train_noma_fair_maml.py',
         '--mat','ISAC_RIS_NOMA_channels_v3_fair.mat',
         '--epochs',EPOCHS,'--batch',b,'--lr',lr,'--w_c','0.7','--w_s','0.3',
         '--out',out,'--iter_log',f'fig4/fig4_b{b}_lr{lr}_iters.npz'])
```

## fig5 — N=16,32,64

```python
for N in (16,32,64):
    run(['python','-u','train_noma_fair_maml.py',
         '--mat',f'ISAC_RIS_NOMA_channels_v3_fair_N{N}.mat',
         '--epochs',EPOCHS,'--batch',BATCH,'--lr',LR,
         '--out',f'fig5/N{N}_b{BATCH}_lr{LR}.pt'])
```

## fig8 — P_max sweep (no-RIS + RIS)

```python
P_list=[5,10,15,20,25]   # low-SINR operating point centred on 15 dBm
for P in P_list:
    run(['python','-u','train_noma_fair_noris.py',
         '--mat','no_ris/ISAC_NOMA_channels_fair_noris.mat',
         '--epochs',60,'--batch',128,'--P_tot_dBm',P,
         '--out',f'fig8/no_ris/noris_P{P}.pt'])
for P in P_list:
    run(['python','-u','train_noma_fair_maml.py',
         '--mat','ISAC_RIS_NOMA_channels_v3_fair_N16.mat',
         '--epochs',EPOCHS,'--batch',BATCH,'--lr',LR,'--P_tot_dBm',P,
         '--out',f'fig8/ris/ris_P{P}_b{BATCH}_lr{LR}.pt'])
```

## eval + ablation -> eval_noma_fair/ (summary.csv x2 + plots)

```python
run(['python','-u','compare_noma_fair.py',
     '--mat','ISAC_RIS_NOMA_channels_v3_fair_TEST.mat',
     '--out-dir','eval_noma_fair','--ckpt_maml','policy_noma_fair_best.pt'])
```

```python
run(['python','fig5/aggregate_fig5.py','--batch',BATCH,'--lr',LR,'--n8_ckpt','policy_noma_fair_best.pt'])
run(['python','fig8/aggregate_fig8.py','--batch',BATCH,'--lr',LR])
```

## bundle (safe zip in /tmp, never self-zips)

```python
import shutil, os, glob
WORK='/kaggle/working'; stage='/tmp/nf_stage'
shutil.rmtree(stage,ignore_errors=True); os.makedirs(stage)
for d in ['fig4','fig5','fig8','eval_noma_fair','plots']:
    s=os.path.join(WORK,d)
    if os.path.isdir(s): shutil.copytree(s, os.path.join(stage,d))
for f in glob.glob(WORK+'/*.pt')+glob.glob(WORK+'/*.npz'): shutil.copy(f,stage)
shutil.make_archive('/tmp/noma_fair_results','zip',stage)
shutil.move('/tmp/noma_fair_results.zip', os.path.join(WORK,'noma_fair_results.zip'))
print('done', round(os.path.getsize(WORK+'/noma_fair_results.zip')/1e6,1),'MB')
for f in ['eval_noma_fair/summary.csv','eval_noma_fair/ablation/summary.csv',
          'fig5/fig5_results.csv','fig8/fig8_results.csv']:
    print(f,'OK' if os.path.exists(os.path.join(WORK,f)) else 'MISSING')
```
